```python
import streamlit as st
import pandas as pd
import gspread
from google.oauth2.service_account import Credentials

# -------------------------------
# CONFIG
# -------------------------------
st.set_page_config(page_title="Team Data Tool", layout="wide")

# -------------------------------
# HEADER
# -------------------------------
st.title("📊 Team Data Upload Tool")
st.markdown("Easily paste, clean, and upload your data to Google Sheets without touching code.")

# -------------------------------
# INPUT AREA
# -------------------------------
st.subheader("Step 1: Paste Your Data")
user_text = st.text_area("Paste data below (CSV format, tabular text, or rows):", height=200)

col1, col2 = st.columns([1,1])
with col1:
    process_btn = st.button("🚀 Process Data")
with col2:
    clear_btn = st.button("🧹 Clear Text")

if clear_btn:
    user_text = ""
    st.experimental_rerun()

# -------------------------------
# PROCESS DATA
# -------------------------------
if process_btn and user_text.strip() != "":
    try:
        # Convert pasted data into DataFrame
        from io import StringIO
        df = pd.read_csv(StringIO(user_text), sep=None, engine="python")

        st.success("✅ Data successfully parsed!")
        st.dataframe(df, use_container_width=True)

        # -------------------------------
        # UPLOAD TO GOOGLE SHEETS
        # -------------------------------
        st.subheader("Step 2: Upload to Google Sheets")

        sheet_id = st.text_input("Enter Google Sheet ID:")
        tab_name = st.text_input("Enter Tab Name (new or existing):", "Processed_Data")
        upload_btn = st.button("📤 Upload to Google Sheet")

        if upload_btn and sheet_id:
            try:
                # Google Auth
                scopes = ["https://www.googleapis.com/auth/spreadsheets"]
                credentials = Credentials.from_service_account_file("service_account.json", scopes=scopes)
                gc = gspread.authorize(credentials)

                spreadsheet = gc.open_by_key(sheet_id)
                try:
                    worksheet = spreadsheet.worksheet(tab_name)
                    spreadsheet.del_worksheet(worksheet)  # Remove old tab if exists
                except:
                    pass
                worksheet = spreadsheet.add_worksheet(title=tab_name, rows=df.shape[0]+10, cols=df.shape[1]+10)

                worksheet.update([df.columns.values.tolist()] + df.values.tolist())
                st.success(f"✅ Data uploaded successfully to Google Sheet tab **{tab_name}**!")

            except Exception as e:
                st.error(f"❌ Upload failed: {e}")

    except Exception as e:
        st.error(f"❌ Could not process data: {e}")
```
